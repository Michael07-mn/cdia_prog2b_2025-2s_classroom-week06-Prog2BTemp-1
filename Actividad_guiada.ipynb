| Propiedades (`@property`) | Para leer y modificar las notas con validaciones |
| Método `promedio()` | Calcula el promedio y muestra si el estudiante **aprueba** o **reprueba** |

## 🧠 Desafío adicional
- Añade validaciones para que el nombre no sea vacío.  
- Crea un método `__str__` para mostrar el estado del estudiante con formato legible.  
- Implementa una lista de varios estudiantes y muestra su promedio grupal.

```python
class Estudiante:
    def __init__(self, nombre: str, nota1: float, nota2: float):
        if not nombre.strip():
            raise ValueError("El nombre no debe estar vacío.")
        self.__nombre = nombre.strip().title()
        self.__nota1 = nota1 
        self.__nota2 = nota2

    @property
    def nota1(self) -> float:
        return self.__nota1

    @nota1.setter
    def nota1(self, valor: float):
        if 0 <= valor <= 10:
            self.__nota1 = valor
        else:
            raise ValueError("Nota1 no válida, debe estar entre 0 y 10")

    @property
    def nota2(self) -> float:
        return self.__nota2

    @nota2.setter
    def nota2(self, valor: float):
        if 0 <= valor <= 10:
            self.__nota2 = valor
        else:
            raise ValueError("Nota2 no válida, debe estar entre 0 y 10")

    def promedio(self) -> float:
        return (self.__nota1 + self.__nota2) / 2
        
    def estado(self) -> str:
       return "Aprobado" if self.promedio() >= 7 else "Reprobado"
    
    def __str__(self) -> str:
        return f"{self.__nombre}: Promedio {self.promedio():.2f} - {self.estado()}"
```

```python
# Crear lista de estudiantes

grupo = [
    Estudiante("Ana", 7.5, 7.9),
    Estudiante("Juan", 7.2, 5.9),
    Estudiante("Carlos", 8.2, 8.8),
    Estudiante("Arelis", 9.5, 7.9),
    Estudiante("Michael", 8.0, 7.5),
]
# Mostrar el estado de cada estudiante

for estudiante in grupo:
    print(estudiante)
```
