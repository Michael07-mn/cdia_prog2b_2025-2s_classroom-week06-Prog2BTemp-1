Agrega `__repr__` para mostrar una vista más técnica del objeto.

__str__ Aparece solo con el print < objeto >

```python
class Producto:
    def __init__(self, nombre, precio, stock):
        self.nombre = nombre
        self.precio = precio
        self.stock = stock
# IMplementar __str__
    def __str__(self):
        return f"Producto: {self.nombre}, Precio: {self.precio} USD, Stock: {self.stock}"
    
    # Agregar __repr__
    def __repr__(self):
        return f"Producto(nombre={self.nombre!r}, precio={self.precio}, stock={self.stock})"
```

## 🧮 B. Nivel 2 — Sobrecarga de operadores

### ⚙️ Ejercicio 3 — Suma y resta de vectores
Implementa una clase `Vector2D(x, y)` que redefina los operadores `+` y `-` y su representación con `__str__`.


Ejemplo para comparar: Suma de listas

```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        # Suma de vectores: x1 + x2,  y1 + y2
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        # Resta los vectores: x1 - x2, y1 - y2
        return Vector2D(self.x - other.x, self.y - other.y)
    
    def __str__(self):
        return f"({self.x}, {self.y})"
```

```python
# Ejemplo de uso
v1 = Vector2D(7, 4)
v2 = Vector2D(3, 2)

print("v1:", v1)
print("v2:", v2)

print("Suma:", v1 + v2)
print("Resta:", v1 - v2)
```

### Producto Punto

```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        # Suma de vectores: x1 + x2,  y1 + y2
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        # Resta los vectores: x1 - x2, y1 - y2
        return Vector2D(self.x - other.x, self.y - other.y)

    def __mul__(self, other):
        #Producto punto: x1 * x2 + y1 * y2
        if isinstance(other, Vector2D):
            return self.x * other.x + self.y * other.y
        else:
            raise TypeError("El producto solo esta definido entre 2 Vector2D ")

    def __str__(self):
        return f"({self.x}, {self.y})"
```

```python
# Ejemplo de uso
v1 = Vector2D(7, 4)
v2 = Vector2D(3, 2)

print("v1:", v1)
print("v2:", v2)

print("Producto punto:", v1 * v2)
```

### ⚖️ Ejercicio 4 — Comparando objetos
Crea una clase `Caja(peso)` y sobrecarga `==` y `<` para comparar por peso.


```python
class Caja:
    def __init__(self, peso):
        self.peso = peso

    def __eq__(self, other):
        # Igualdad, compara pesos
        return self.peso == other.peso
    
    def __lt__(self, other):
        # Menor que, compara pesos
        return self.peso < other.peso
    def __str__(self):
        return f"Caja(peso={self.peso})"
```

```python
# --- Pruebas ---
c1 = Caja(10)
c2 = Caja(15)
c3 = Caja(25)

print("c1:", c1)
print("c2:", c2)
print("c3:", c3)

print("c1 == c2:", c1 == c2)  # False
print("c1 == c3:", c1 == c3)  # False
print("c1 < c2:", c1 < c2)    # True
print("c2 < c3:", c2 < c3)    # True
```

## 🧰 C. Nivel 3 — Contenedores y otros métodos mágicos

### 📦 Ejercicio 5 — Inventario personalizado
Diseña una clase `Inventario` que se comporte como una lista.

- `__len__` → número de productos  
- `__getitem__` → acceder por índice  
- `__setitem__` → modificar  
- `__contains__` → verificar si un producto está incluido  


```python
class Inventario:
    def __init__(self):
        #Internamente se usa una lista para almacenar los productos
        self._productos = []

    def __len__(self):
        # DEvuelve el número de elementos que hay en la lista
        return len(self._productos)

    def __getitem__(self, index):
        # Permite acceder al inventario[i]
        return self._productos[index]
    
    def __setitem__(self, index, value):
        # Permite modificar el inventario.
        self._productos[index] = value

    def __contains__(self, item):
        # Permite usar 'in' para verificar si un producto está incluido
        return item in self._productos
    
    def agregar(self, producto):
        # Método auxiliar para agregar productos
        self._productos.append(producto)

    def __repr__(self):
        # Representación amigable del inventario
        return f"Inventario: {self._productos}"
```

```python
#Uso
inv = Inventario()
inv.agregar("Manzanas")
inv.agregar("Peras")
inv.agregar("Uvas")
inv.agregar("Naranjas")
print(inv)

print("Número de productos:", len(inv))   # Usa __len__
print("Primer producto:", inv[0])         # Usa __getitem__
inv[1] = "Bananas"                        # Usa __setitem__
print(inv)                                # Inventario: ['Manzanas', 'Bananas', 'Uvas', 'Naranjas']
print("Babacos" in inv)                     # Usa __contains__ → False
print("Uvas" in inv)                   # True
```

## 🧠 D. Nivel 4 — Patrón Singleton

### 🧩 Ejercicio 6 — Gestor de configuración única
Asegura que solo exista **una instancia** de la clase `Configuracion` usando `__new__`.


```python
class Configuracion:
    _instancia = None # Atributo de clase para guardar la única instancia

    def __new__(cls, *args, **kwargs):
        # Si no existe, se crea y se guarda en _instancia.
        if cls._instancia is None:
            cls._instancia = super().__new__(cls)
        return cls._instancia
    
    def __init__(self, modo="producción"):
        #Este puede ser llamada varias veces, pero la instancia es única

        self.modo = modo

    def __str__(self):
        return f"Configuración única (modo={self.modo})"
```

```python
c1 = Configuracion("desarrollo")
c2 = Configuracion("producción")

print(c1)  # Configuración única (modo=producción)
print(c2)  # Configuración única (modo=producción)

print("¿c1 es c2?", c1 is c2)  # True
```

### 🔁 Ejercicio 7 — Contador Singleton
Crea una clase `Contador` que acumule valores con `+` sin crear nuevas instancias.


```python
class Contador:
    _instancia = None  # atributo de clase para guardar la única instancia

    def __new__(cls, *args, **kwargs):
        # Garantiza que solo exista una instancia
        if cls._instancia is None:
            cls._instancia = super().__new__(cls)
            cls._instancia.valor = 0  # inicializa el contador
        return cls._instancia

    def __add__(self, other):
        # Sobrecarga del operador + para acumular
        if isinstance(other, (int, float)):
            self.valor += other
            return self  # devuelve la misma instancia
        raise TypeError("Solo se puede sumar números al Contador")

    def __str__(self):
        return f"Contador(valor={self.valor})"
```

```python
c1 = Contador()
c2 = Contador()

print("¿c1 es c2?", c1 is c2)  # True, misma instancia

c1 + 5
print(c1)  # Contador(valor=5)

c2 + 10
print(c2)  # Contador(valor=15) → mismo objeto acumulado
```

### 🏦 Ejercicio 8 — Sistema de cuentas bancarias
Diseña una clase `CuentaBancaria` con:

- `__init__`: nombre y saldo  
- `__str__`: mostrar el saldo  
- `__add__`: combinar cuentas  
- `__sub__`: transferir dinero  
- `__eq__`: comparar saldos  

**Nota:** Puedes agregar validaciones básicas (por ejemplo, evitar saldos negativos en transferencias).


```python
class CuentaBancaria:
    def __init__(self, nombre, saldo=0):
        self.nombre = nombre
        self.saldo = saldo

    def __str__(self):
        # Representación amigable
        return f"Cuenta de {self.nombre} → Saldo: {self.saldo} USD"

    def __add__(self, other):
        # Combinar cuentas: crea una nueva con la suma de saldos
        if isinstance(other, CuentaBancaria):
            nuevo_nombre = f"{self.nombre}&{other.nombre}"
            nuevo_saldo = self.saldo + other.saldo
            return CuentaBancaria(nuevo_nombre, nuevo_saldo)
        raise TypeError("Solo se pueden combinar cuentas bancarias")

    def __sub__(self, other):
        # Transferir dinero: resta saldo de self y suma a other
        if isinstance(other, CuentaBancaria):
            monto = min(self.saldo, 100)  # ejemplo: transferir hasta 100
            if monto <= 0:
                raise ValueError("No hay fondos suficientes para transferir")
            self.saldo -= monto
            other.saldo += monto
            return f"Transferidos {monto} USD de {self.nombre} a {other.nombre}"
        raise TypeError("Solo se pueden transferir entre cuentas bancarias")

    def __eq__(self, other):
        # Comparar saldos
        if isinstance(other, CuentaBancaria):
            return self.saldo == other.saldo
        return False
```

```python
c1 = CuentaBancaria("Ana", 500)
c2 = CuentaBancaria("Carlos", 300)
c3 = CuentaBancaria("Luis", 500)

print(c1)  # Cuenta de Ana → Saldo: 500 USD
print(c2)  # Cuenta de Carlos → Saldo: 300 USD

# Combinar cuentas
c4 = c1 + c2
print(c4)  # Cuenta de Ana&Carlos → Saldo: 800 USD

# Transferencia
print(c1 - c2)  # Transferidos 100 USD de Ana a Carlos
print(c1)       # Saldo actualizado
print(c2)       # Saldo actualizado

# Comparación de saldos
print("c1 == c3:", c1 == c3)  # True si tienen el mismo saldo
```
