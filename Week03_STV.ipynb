## Ejercicio 1 — Empleado / Gerente (override + `super()`)
**Enunciado:**  
Diseña una jerarquía con `Empleado(nombre, salario)` y `Gerente(nombre, salario, depto)` donde `Gerente` **sobrescriba** un método `mostrar_info()` y **reutilice** el constructor de la clase base mediante `super()`.

**Requerimientos mínimos:**
- `Empleado` debe almacenar nombre y salario, y contar con `mostrar_info()`.
- `Gerente` debe heredar de `Empleado`, añadir `depto` y sobrescribir `mostrar_info()`.
- Preparar **al menos 2 instancias** y mostrar su salida (polimorfismo).

**Objetivo de aprendizaje:** comprender **herencia simple**, **overriding** y uso de `super()`.


```python
#Clase Base
class Empleado:
    def __init__(self, nombre, salario): #atributos nombre y salario
        self.nombre = nombre
        self.salario = salario

# Método de la clase Base
    def mostrar_info(self):
        return f"Empleado: {self.nombre}, Salario: {self.salario}"

class Gerente(Empleado):
    def __init__(self, nombre, salario, depto):
        #Reusar el constructor de la clase base
        super().__init__(nombre, salario)
        self.depto = depto

    def mostrar_info(self):
        # Sobreescribir el método
        base_info = super().mostrar_info()
        return f"{base_info}, Departamento: {self.depto}"
```

```python
# Instancias
e_1 = Empleado("Juan", 1740)
g_1 = Gerente("Martín", 800, "Finanzas")
e_2 = Empleado("Rashel", 1400)
g_2 = Gerente("Michael", 2600, "Data Science")

# Polimorfismo
for p in [e_1, g_1, e_2, g_2]:
    print(p.mostrar_info())
```

## Ejercicio 2 — Animal / Perro / Gato (polimorfismo en colección)
**Enunciado:**  
Implementa `Animal.hablar()` y dos derivados (`Perro`, `Gato`). Recorre una **colección heterogénea** e invoca `hablar()` a cada elemento para observar el comportamiento polimórfico.

**Requerimientos mínimos:**
- Clase base `Animal` con un método `hablar()` (puede ser abstracto o con implementación genérica).

## Ejercicio 3 — Figuras (clase abstracta + métodos abstractos)
**Enunciado:**  
Define una clase abstracta `Figura` con métodos `area()` y `perimetro()` y crea implementaciones `Rectangulo(w,h)` y `Circulo(r)`. Calcula el **área total** de una lista de figuras.

**Requerimientos mínimos:**
- `Figura` como clase abstracta con `area()` y `perimetro()` abstractos.
- Implementaciones concretas `Rectangulo` y `Circulo`.
- Lista de `Figura` y suma de áreas con comprensión o bucle.

**Objetivo de aprendizaje:** diferenciar **interfaces/abstractas** de **implementaciones** y usar polimorfismo para **agregación** de resultados.


```python
from abc import ABC, abstractmethod
import math

# Clase abstracta
class Figura(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimetro(self):
        pass

# Implementaciones
class Rectangulo(Figura):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h
    
    def perimetro(self):
        return 2 * (self.w + self.h)
    
class Circulo(Figura):
    def __init__(self, r):
        self.r = r

    def area(self):
        return math.pi * self.r**2
    
    def perimetro(self):
        return 2 * math.pi * self.r
```

```python
figuras = [
    Rectangulo(4, 6),
    Circulo(4),
    Rectangulo(2, 8)
]
# Calcular el área total
area_total = sum(fig.area() for fig in figuras)

# Resultados
for fig in figuras:
    print(f"{fig.__class__.__name__} -> Area: {fig.area():.2f}, Perímetro: {fig.perimetro():.2f}")
print(f"\nÁrea total de todas las figuras: {area_total:.2f}")
```

## Ejercicio 4 — Múltiple herencia / Mixin de Logger
**Enunciado:**  
Crea un mixin `LoggerMixin` con un método `log(msg)` y una clase `Servicio` con `procesar(x)`. Diseña `ServicioConLog` que **herede** de ambos y registre el flujo antes y después de procesar.

**Requerimientos mínimos:**
- `LoggerMixin.log(msg)` que imprima o almacene mensajes.
- `Servicio.procesar(x)` que transforme el dato (p. ej., `x*2`).
- `ServicioConLog` que use el mixin para añadir trazas sin duplicar lógica.
- Explica brevemente el **orden de resolución de métodos (MRO)** implicado.

**Objetivo de aprendizaje:** comprender **herencia múltiple** (o mixins) y **separación de responsabilidades**.


```python
class LoggerMixin:
    def log(self, msg):
        print(f"[LOG]: {msg}")

class Servicio:
    def procesar(self, x):
        return x * 2

class ServicioConLog(LoggerMixin, Servicio):
    def procesar(self, x):
        # Antes de procesar
        self.log(f"Iniciando procesamiento de {x}")
        
        resultado = super().procesar(x)  # reutiliza lógica de Servicio
        
        # Después de procesar
        self.log(f"Finalizó procesamiento, resultado = {resultado}")
        return resultado
```

```python
# Uso
servicio = ServicioConLog()
print("Resultado final:", servicio.procesar(10))

"""El MRO (Method Resolution Order) es la forma en que Python decide qué método
ejecutar primero cuando hay herencia múltiple."""
```

## Ejercicio 5 — Pagos (polimorfismo + fábrica simple)
**Enunciado:**  
Diseña una jerarquía `Pago` con `procesar()` e implementaciones `PagoEfectivo` (sin comisión) y `PagoTarjeta` (con `fee%`). Crea una función que reciba una lista heterogénea de pagos y devuelva los montos procesados.

**Requerimientos mínimos:**
- Clase base `Pago(monto)` con método `procesar()` (abstracto o genérico).
- Subclases `PagoEfectivo` y `PagoTarjeta(fee)` que **sobrescriban** `procesar()`.
- Función `ejecutar_pagos(pagos)` que aplique polimorfismo.
- (Opcional) Una **fábrica** simple para crear el tipo de pago a partir de un diccionario de configuración.

**Objetivo de aprendizaje:** practicar **polimorfismo** en una operación de negocio y **extensibilidad** mediante fábrica.


```python
from abc import ABC, abstractmethod

# --- Clase base abstracta ---
class Pago(ABC):
    def __init__(self, monto):
        self.monto = monto

    @abstractmethod
    def procesar(self):
        pass

# --- Implementaciones concretas ---
class PagoEfectivo(Pago):
    def procesar(self):
        # Sin comisión
        return self.monto

class PagoTarjeta(Pago):
    def __init__(self, monto, fee):
        super().__init__(monto)
        self.fee = fee  # porcentaje de comisión

    def procesar(self):
        # Aplica comisión
        return self.monto * (1 + self.fee / 100)

# --- Función polimórfica ---
def ejecutar_pagos(pagos):
    return [pago.procesar() for pago in pagos]

# --- Fábrica simple ---
def crear_pago(config):
    tipo = config.get("tipo")
    monto = config.get("monto")
    if tipo == "efectivo":
        return PagoEfectivo(monto)
    elif tipo == "tarjeta":
        fee = config.get("fee", 0)
        return PagoTarjeta(monto, fee)
    else:
        raise ValueError(f"Tipo de pago desconocido: {tipo}")
```

```python
# USO
configs = [
    {"tipo": "efectivo", "monto": 500},
    {"tipo": "tarjeta", "monto": 200, "fee": 10},
    {"tipo": "tarjeta", "monto": 150, "fee": 2},
]

# Crear instancias con la fábrica
pagos = [crear_pago(cfg) for cfg in configs]

# Ejecutar polimórficamente
resultados = ejecutar_pagos(pagos)

# --- Mostrar resultados ---
for pago, resultado in zip(pagos, resultados):
    print(f"{pago.__class__.__name__} -> Procesado: {resultado:.2f}")
```
