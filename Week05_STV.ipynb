## 6. Indexación y slicing

**Ejercicio 6.1** → Dada la matriz `A = [[10, 20, 30],[40, 50, 60]]`, extrae: (a) el elemento 50, (b) la primera fila, (c) la tercera columna.

```python
# Tu código aquí
A = np.array([[10, 20, 30],
              [40, 50, 60]])

# elemento 50 -> fila 1, columna 1
el_50 = A[1,1]

# primera finally
fila_1 = A[0, :]

# 3ra columna
col_3 = A[:, 2]

print("(a) elemento 50:", el_50)
print("(b) primera fila:", fila_1)
print("(c) tercera columna:", col_3)

```

**Ejercicio 6.2** → Usando slicing sobre `np.arange(20)`, obtén los elementos en posiciones pares (índices 0,2,4,...) y los últimos 5 elementos.

```python
# Tu código aquí
v = np.arange(20)
pares = v[::2]
ult_5 = v[-5:]

print("Vector original:", v)
print("Posiciones pares:", pares)
print("Últimos 5 elementos:", ult_5)
```
